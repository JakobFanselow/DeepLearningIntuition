```python
import torch, torch.nn as nn
import matplotlib.pyplot as plt

torch.manual_seed(0)

true_w, true_b = 2.0, 1.0
x = torch.rand(100, 1) * 10
y = true_w * x + true_b + torch.randn_like(x) * 2.0

plt.figure(figsize=(8, 6))

plt.scatter(x.numpy(), y.numpy(), color='royalblue', alpha=0.7, edgecolors='k', label='Data Points')

x_line = torch.linspace(0, 10, 100).unsqueeze(1)
y_line = (true_w+0.05) * (x_line) + (true_b)
plt.plot(x_line.numpy(), y_line.numpy(), color='crimson', linewidth=2.5, label=f'Proposal 1')

x_line = torch.linspace(0, 10, 100).unsqueeze(1)
y_line = (true_w-0.1) * (x_line) + (true_b+0.4)
plt.plot(x_line.numpy(), y_line.numpy(), color='green', linewidth=2.5, label=f'Proposal 2')

plt.xlabel('x', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
